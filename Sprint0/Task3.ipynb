```
// =========================================================================
// 1. БАЗОВЫЙ КЛАСС (Employee)
// =========================================================================
public class Employee
{
    // Обязательные базовые атрибуты
    public string Name { get; set; } = string.Empty;
    public int Age { get; set; }
    public decimal Salary { get; set; }

    // Дополнительные атрибуты (3-4 атрибута)
    public string Department { get; set; } = string.Empty;
    public DateTime HireDate { get; set; } = DateTime.Now;
    public decimal Bonus { get; set; }
    public string Email { get; set; } = string.Empty;

    // Конструкторы
    public Employee() { }

    public Employee(string name, int age, decimal salary)
    {
        Name = name;
        Age = age;
        Salary = salary;
        Email = $"{name.ToLower().Replace(" ", ".")}@company.com";
    }

    public Employee(string name, int age, decimal salary, string department)
        : this(name, age, salary)
    {
        Department = department;
    }

    public Employee(string name, int age, decimal salary, string department, decimal bonus, string email)
        : this(name, age, salary, department)
    {
        Bonus = bonus;
        Email = email;
    }

    // Базовые методы
    public virtual string GetInfo()
    {
        string dept = string.IsNullOrWhiteSpace(Department) ? "Общий отдел" : Department;
        return $"Имя: {Name} | Возраст: {Age} | Зарплата: {Salary:N0} руб. | Отдел: {dept} | Премия: {Bonus:N0} руб.";
    }

    public virtual void Work()
    {
        Console.WriteLine($"Сотрудник {Name} выполняет базовые рабочие обязанности");
    }

    public virtual void TakeVacation()
    {
        Console.WriteLine($"Сотрудник {Name} уходит в стандартный оплачиваемый отпуск");
    }

    // Дополнительные методы (3-4 метода)
    public void CalculateBonus(decimal percentage)
    {
        Bonus = Salary * (percentage / 100m);
        Console.WriteLine($"Сотруднику {Name} начислена премия {percentage}% ({Bonus:N0} руб.). Итого к выплате: {Salary + Bonus:N0} руб.");
    }

    public void IncreaseSalary(decimal amount)
    {
        Salary += amount;
        Console.WriteLine($"Зарплата сотрудника {Name} увеличена на {amount:N0} руб. Новая ставка: {Salary:N0} руб.");
    }

    public void ChangeDepartment(string newDepartment)
    {
        Console.WriteLine($"Сотрудник {Name} переведен из отдела '{Department}' в '{newDepartment}'");
        Department = newDepartment;
    }

    public virtual void Dismiss(string reason)
    {
        Console.WriteLine($"Сотрудник {Name} уволен. Причина: {reason}");
    }
}

// =========================================================================
// 2. ПРОСТОЕ НАСЛЕДОВАНИЕ (Single Inheritance)
// =========================================================================

// --- Класс 1: Менеджер (Manager) ---
public class Manager : Employee
{
    // Обязательный атрибут
    public int SubordinatesCount { get; set; }

    // Дополнительные атрибуты (3-4 атрибута)
    public decimal TeamBudget { get; set; }
    public string ManagementLevel { get; set; } = "Middle";
    public int CompletedProjectsCount { get; set; }

    // Конструкторы
    public Manager() { }

    public Manager(string name, int age, decimal salary, int subordinatesCount)
        : base(name, age, salary)
    {
        SubordinatesCount = subordinatesCount;
    }

    public Manager(string name, int age, decimal salary, string department, int subordinatesCount, decimal teamBudget, string managementLevel)
        : base(name, age, salary, department)
    {
        SubordinatesCount = subordinatesCount;
        TeamBudget = teamBudget;
        ManagementLevel = managementLevel;
    }

    // Переопределенные методы (полиморфизм)
    public override void Work()
    {
        base.Work();
        Console.WriteLine($"Менеджер {Name} управляет командой из {SubordinatesCount} человек");
    }

    public override string GetInfo()
    {
        string baseInfo = base.GetInfo();
        return $"{baseInfo} | Подчиненных: {SubordinatesCount} | Уровень: {ManagementLevel} | Бюджет: {TeamBudget:N0} руб.";
    }

    // Дополнительные методы
    public void ConductMeeting()
    {
        Console.WriteLine($"Менеджер {Name} проводит рабочее совещание с командой");
    }

    public void ApproveBudget(decimal amount)
    {
        TeamBudget += amount;
        Console.WriteLine($"Менеджер {Name} утвердил дополнительный бюджет в размере {amount:N0} руб. Текущий бюджет: {TeamBudget:N0} руб.");
    }

    public void AssignBonusToSubordinate(Employee employee, decimal amount)
    {
        employee.Bonus += amount;
        Console.WriteLine($"Менеджер {Name} выписал премию {amount:N0} руб. сотруднику {employee.Name}");
    }
}

// --- Класс 2: Разработчик (Developer) ---
public class Developer : Employee
{
    // Обязательный атрибут
    public string Specialization { get; set; } = string.Empty;

    // Дополнительные атрибуты (3-4 атрибута)
    public string ProgrammingLanguage { get; set; } = "C#";
    public string GitHubUsername { get; set; } = string.Empty;
    public int ClosedTicketsCount { get; set; }

    // Конструкторы
    public Developer() { }

    public Developer(string name, int age, decimal salary, string specialization)
        : base(name, age, salary)
    {
        Specialization = specialization;
    }

    public Developer(string name, int age, decimal salary, string department, string specialization, string programmingLanguage, string gitHubUsername)
        : base(name, age, salary, department)
    {
        Specialization = specialization;
        ProgrammingLanguage = programmingLanguage;
        GitHubUsername = gitHubUsername;
    }

    // Переопределенные методы (полиморфизм)
    public override void TakeVacation()
    {
        base.TakeVacation();
        Console.WriteLine($"Разработчик {Name} ({Specialization}) должен согласовать отпуск с проектными сроками");
    }

    public override string GetInfo()
    {
        string baseInfo = base.GetInfo();
        return $"{baseInfo} | Специализация: {Specialization} | Язык: {ProgrammingLanguage} | GitHub: {GitHubUsername}";
    }

    // Дополнительные методы
    public void WriteCode()
    {
        Console.WriteLine($"Разработчик {Name} разрабатывает функционал по направлению {Specialization}");
    }

    public void PerformCodeReview(string pullRequestTitle)
    {
        Console.WriteLine($"Разработчик {Name} провел ревью пулл-реквеста '{pullRequestTitle}'");
    }

    public void ResolveTicket(string ticketId)
    {
        ClosedTicketsCount++;
        Console.WriteLine($"Разработчик {Name} успешно закрыл тикет {ticketId} (Всего закрыто: {ClosedTicketsCount})");
    }
}

// --- Класс 3: Дизайнер (Designer) ---
public class Designer : Employee
{
    // Обязательный атрибут
    public string DesignStyle { get; set; } = string.Empty;

    // Дополнительные атрибуты (3-4 атрибута)
    public string PrimaryTool { get; set; } = "Figma";
    public string PortfolioUrl { get; set; } = string.Empty;
    public int MockupsCreated { get; set; }

    // Конструкторы
    public Designer() { }

    public Designer(string name, int age, decimal salary, string designStyle)
        : base(name, age, salary)
    {
        DesignStyle = designStyle;
    }

    public Designer(string name, int age, decimal salary, string department, string designStyle, string primaryTool, string portfolioUrl)
        : base(name, age, salary, department)
    {
        DesignStyle = designStyle;
        PrimaryTool = primaryTool;
        PortfolioUrl = portfolioUrl;
    }

    // Переопределенные методы (полиморфизм)
    public override void Work()
    {
        base.Work();
        Console.WriteLine($"Дизайнер {Name} работает над дизайн-проектом в стиле '{DesignStyle}'");
    }

    public override string GetInfo()
    {
        string baseInfo = base.GetInfo();
        return $"{baseInfo} | Стиль дизайна: {DesignStyle} | Инструмент: {PrimaryTool} | Портфолио: {PortfolioUrl}";
    }

    // Дополнительные методы
    public void CreatePrototype()
    {
        MockupsCreated++;
        Console.WriteLine($"Дизайнер {Name} создал интерактивный прототип проекта в стиле '{DesignStyle}' в {PrimaryTool} (Макетов: {MockupsCreated})");
    }

    public void PresentDesign(string clientName)
    {
        Console.WriteLine($"Дизайнер {Name} презентовал концепт-дизайн клиенту '{clientName}'");
    }
}

// =========================================================================
// 3. СЛОЖНОЕ (МНОГОУРОВНЕВОЕ) НАСЛЕДОВАНИЕ (Multi-level Inheritance)
// Иерархия: Employee -> Developer -> TechLead
// =========================================================================

public class TechLead : Developer
{
    public string ManagedTeam { get; set; } = string.Empty;
    public string TechnicalVision { get; set; } = string.Empty;
    public int ArchitectureLevel { get; set; } = 5;

    public TechLead() { }

    public TechLead(string name, int age, decimal salary, string specialization, string managedTeam, string technicalVision)
        : base(name, age, salary, specialization)
    {
        ManagedTeam = managedTeam;
        TechnicalVision = technicalVision;
    }

    public TechLead(string name, int age, decimal salary, string department, string specialization, string programmingLanguage, string gitHubUsername, string managedTeam, string technicalVision, int architectureLevel)
        : base(name, age, salary, department, specialization, programmingLanguage, gitHubUsername)
    {
        ManagedTeam = managedTeam;
        TechnicalVision = technicalVision;
        ArchitectureLevel = architectureLevel;
    }

    public override void Work()
    {
        base.Work();
        Console.WriteLine($"[TECH LEAD] Техлид {Name} руководит техстеком команды '{ManagedTeam}' и развивает архитектуру: '{TechnicalVision}'");
    }

    public override string GetInfo()
    {
        return $"{base.GetInfo()} | Команда: {ManagedTeam} | Уровень архитектора: {ArchitectureLevel}/5 | Видение: {TechnicalVision}";
    }

    public void ApproveArchitecture(string solutionName)
    {
        Console.WriteLine($"[АРХИТЕКТУРА] Техлид {Name} утвердил архитектурное решение: '{solutionName}'");
    }

    public void ConductTechInterview(string candidateName)
    {
        Console.WriteLine($"[ИНТЕРВЬЮ] Техлид {Name} проводит техническое собеседование с кандидатом: '{candidateName}'");
    }
}

// =========================================================================
// 4. МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ ЧЕРЕЗ ИНТЕРФЕЙСЫ (Multiple via Interfaces)
// =========================================================================

public interface INotifiable
{
    string NotificationChannel { get; set; }
    void SendNotification(string message);
}

public interface IReportable
{
    DateTime LastReportedDate { get; set; }
    string GenerateReport();
}

public class DepartmentDirector : Manager, INotifiable, IReportable
{
    public string NotificationChannel { get; set; } = "Корпоративный Telegram / Email";
    public DateTime LastReportedDate { get; set; } = DateTime.Now;
    public string StrategicDirection { get; set; } = string.Empty;

    public DepartmentDirector() { }

    public DepartmentDirector(string name, int age, decimal salary, int subordinatesCount, string strategicDirection, string notificationChannel)
        : base(name, age, salary, subordinatesCount)
    {
        StrategicDirection = strategicDirection;
        NotificationChannel = notificationChannel;
    }

    public DepartmentDirector(string name, int age, decimal salary, string department, int subordinatesCount, decimal teamBudget, string managementLevel, string strategicDirection, string notificationChannel)
        : base(name, age, salary, department, subordinatesCount, teamBudget, managementLevel)
    {
        StrategicDirection = strategicDirection;
        NotificationChannel = notificationChannel;
    }

    public void SendNotification(string message)
    {
        Console.WriteLine($"[Уведомление в {NotificationChannel}] Директор {Name}: {message}");
    }

    public string GenerateReport()
    {
        LastReportedDate = DateTime.Now;
        return $"[СТРАТЕГИЧЕСКИЙ ОТЧЕТ] Директор: {Name} | Отдел: {Department} | Подчиненных: {SubordinatesCount} чел. | Направление: {StrategicDirection} | Дата отчета: {LastReportedDate:dd.MM.yyyy HH:mm}";
    }

    public void ExecuteStrategicInitiative(string initiativeName)
    {
        Console.WriteLine($"[СТРАТЕГИЯ] Директор {Name} запустил инициативу: '{initiativeName}'");
    }
}

// =========================================================================
// ДЕМОНСТРАЦИЯ ВСЕХ ВИДОВ НАСЛЕДОВАНИЯ
// =========================================================================

Console.WriteLine("=================================================");
Console.WriteLine("=== 1. ПРОСТОЕ НАСЛЕДОВАНИЕ (Single Inheritance) ===");
Console.WriteLine("=================================================");

Employee myEmployee = new Employee("Иван Иванов", 30, 50000, "Операционный отдел");
Manager myManager = new Manager("Ольга Смирнова", 38, 120000, 7);
Developer myDeveloper = new Developer("Алексей Попов", 26, 95000, "Backend C#");
Designer myDesigner = new Designer("Мария Соколова", 24, 80000, "UI/UX & Minimal");

// Базовый сотрудник
Console.WriteLine(myEmployee.GetInfo());
myEmployee.Work();
myEmployee.CalculateBonus(15);
myEmployee.TakeVacation();

Console.WriteLine();
// Менеджер
Console.WriteLine(myManager.GetInfo());
myManager.Work();
myManager.ConductMeeting();
myManager.ApproveBudget(150000);

Console.WriteLine();
// Разработчик
Console.WriteLine(myDeveloper.GetInfo());
myDeveloper.Work();
myDeveloper.WriteCode();
myDeveloper.PerformCodeReview("PR-104: Авторизация OAuth2");
myDeveloper.TakeVacation();

Console.WriteLine();
// Дизайнер
Console.WriteLine(myDesigner.GetInfo());
myDesigner.Work();
myDesigner.CreatePrototype();
myDesigner.PresentDesign("Заказчик ООО Технологии");

Console.WriteLine("\n========================================================");
Console.WriteLine("=== 2. СЛОЖНОЕ НАСЛЕДОВАНИЕ (Multi-level Inheritance) ===");
Console.WriteLine("========================================================");

TechLead myTechLead = new TechLead(
    "Богунова Екатерина",
    28,
    180000,
    "Fullstack & Cloud Architecture",
    "HITS Core Team",
    "Микросервисы на .NET 10 и gRPC"
);

Console.WriteLine(myTechLead.GetInfo());
myTechLead.Work();
myTechLead.WriteCode();
myTechLead.ApproveArchitecture("Переход на событийную шину Kafka");
myTechLead.ConductTechInterview("Дмитрий Кузнецов (Middle C#)");
myTechLead.CalculateBonus(25);
myTechLead.TakeVacation();

Console.WriteLine("\n============================================================");
Console.WriteLine("=== 3. МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ ЧЕРЕЗ ИНТЕРФЕЙСЫ ===");
Console.WriteLine("============================================================");

DepartmentDirector director = new DepartmentDirector(
    "Сергей Воронов",
    45,
    250000,
    "Департамент ИТ-инноваций",
    45,
    5000000,
    "Top Executive",
    "Цифровая трансформация и AI-интеграция",
    "Telegram Корпоративный"
);

Console.WriteLine(director.GetInfo());
director.Work();
director.ConductMeeting();
director.SendNotification("Старт спринта по внедрению LLM-ассистентов в рабочие процессы");
director.ExecuteStrategicInitiative("Автоматизация внутренних бизнес-процессов компании");
Console.WriteLine(director.GenerateReport());
```
